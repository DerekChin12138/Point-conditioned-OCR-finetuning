# 01 · 准星协议与提示词

训练图和推理图必须用同一套 `MARKER_SPEC` 画准星。


```python
from pathlib import Path
import os, sys
ROOT = Path("..").resolve() if Path.cwd().name == "notebooks" else Path.cwd().resolve()
os.chdir(ROOT)
sys.path.insert(0, str(ROOT / "src"))

from PIL import Image, ImageDraw
from IPython.display import display
from point_ocr.marker import MARKER_SPEC, draw_crosshair
from point_ocr.prompts import POINT_PROMPT, PAGE_PROMPT

print("MARKER_SPEC:", MARKER_SPEC)
print("--- POINT_PROMPT (前 200 字) ---")
print(POINT_PROMPT[:200], "...")
```

```python
# 造一张假页面，叠准星
img = Image.new("RGB", (800, 500), (245, 245, 240))
d = ImageDraw.Draw(img)
d.rectangle([60, 50, 520, 120], outline=(0, 0, 0), fill=(255, 255, 255))
d.text((80, 70), "Heading block under crosshair", fill=(0, 0, 0))
d.rectangle([60, 160, 700, 320], outline=(0, 0, 0), fill=(255, 255, 255))
d.text((80, 200), "Body paragraph that should NOT be dumped", fill=(0, 0, 0))

marked = draw_crosshair(img, x=200, y=85)
out = ROOT / "notebooks" / "_demo_marked.jpg"
out.parent.mkdir(exist_ok=True)
marked.save(out)
print("saved", out)
display(marked.resize((400, 250)))
```

观察：品红十字 + 白环应非常醒目。产品侧以后也必须调用同一 `draw_crosshair`。
